<h2 style="color:#FFFFFF; text-align:center; font-family:'Roboto', sans-serif; font-size:24px; font-weight:600; letter-spacing:0.08em;">INTEGRATING RAZER AIKIT WITH OPENAI API</h2>

<h3 style="color:#44D62C; text-align:left;">Objectives of This Guide</h3>

This guide demonstrates how to use the official <strong>OpenAI Python SDK</strong> to perform inference on language models deployed through <strong>Razer AIKit</strong>. Because AIKit exposes model serving through <strong>OpenAI-compatible API endpoints</strong>, developers can seamlessly connect the <code>openai</code> library to local models without modifying existing codebases built for OpenAI's cloud service.

By following this guide, you will learn how to:

<ul>
  <li>Configure the OpenAI Python SDK</li>
  <li>Point the OpenAI client to a AIKit-hosted local server</li>
  <li>Send chat or completion requests to models running on-device via AIKit>
  <li>Validate responses and ensure compatibility with existing OpenAI-style workflows</li>
</ul>


<h3 style="color:#44D62C; text-align:left;">📥 1. Check Compatibility & Download a Model</h3>

Use these commands to check compatibility and pull a model from Hugging Face into your local environment.  
This ensures it's available for fast and offline inference.

```python
!rzr-aikit model info ibm-granite/granite-3.3-2b-instruct
```

```python
!rzr-aikit model download ibm-granite/granite-3.3-2b-instruct
```

<h3 style="color:#44D62C; text-align:left;">🚀 2. Run a Model</h3>

Start a model server to perform inference locally.

When you run this command, `rzr-aikit` automatically applies optimized configuration for the selected model and your hardware (e.g., dtype, max tokens, batch size, etc.). However, you can always provide your configuration as an options.

```python
!nohup rzr-aikit model run ibm-granite/granite-3.3-2b-instruct
# Output is available in nohup.out
```

<h3 style="color:#44D62C; text-align:left;">💬 3. OpenAI API: Basic Chat Completion</h3>

Once the model is running, you can send prompts and receive completions using the `generate` command.

This is ideal for interactive testing and quick validation of the model’s behavior.

```python
from openai import OpenAI

client = OpenAI(
    base_url="http://localhost:8000/v1",  # AIKit
    api_key="rzr-aikit"  # required by SDK, but not validated
)

completion = client.chat.completions.create(
    model="ibm-granite/granite-3.3-2b-instruct",
    messages=[
        {"role": "user", "content": "What is the capital of Japan?"}
    ]
)

print(completion.choices[0].message.content)
```

<h3 style="color:#44D62C; text-align:left;">🧠 4. OpenAI API: System Prompt + Multi-Turn Chat</h3>

This example demonstrates a more advanced use of the API, where multiple messages — including a system prompt — define a conversation context.
Useful for role-playing, chatbots, and multi-turn interactions.

```python
messages = [
    {"role": "system", "content": "You are a sarcastic tech support agent."},
    {"role": "user", "content": "My PC won't turn on."},
    {"role": "assistant", "content": "Did you try threatening it with a hammer?"},
    {"role": "user", "content": "That's not helpful."}
]

completion = client.chat.completions.create(
    model="ibm-granite/granite-3.3-2b-instruct",
    messages=messages
)

print(completion.choices[0].message.content)
```

<h3 style="color:#44D62C; text-align:left;">⚡ 5. OpenAI API: Streaming Response</h3>

This example enables streaming output from the model token-by-token as it generates text.
Great for real-time experiences like terminals, chat UIs, and voice pipelines.

```python
stream = client.chat.completions.create(
    model="ibm-granite/granite-3.3-2b-instruct",
    messages=[
        {"role": "user", "content": "Write a short poem about cyberpunk cats."}
    ],
    stream=True,
)

for chunk in stream:
    print(chunk.choices[0].delta.content or "", end="")
```
