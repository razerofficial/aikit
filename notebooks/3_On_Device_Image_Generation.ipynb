<h2 style="color:#FFFFFF; text-align:center; font-family:'Roboto', sans-serif; font-size:24px; font-weight:600; letter-spacing:0.08em;">GUIDE FOR USING DIFFUSION MODELS FOR LOCAL IMAGE GENERATION.</h2>

<h3 style="color:#44D62C; text-align:left;">Objectives of This Guide</h3>

This guide is designed to walk you through the end-to-end process of generating images locally using diffusion models via the `rzr-aikit` interface.

<h3 style="color:#44D62C; text-align:left;">Project Overview</h3>

AIKit is Razer's AI developer environment built to simplify and accelerate machine learning workflows on high-performance Razer hardware. It offers a plug-and-play experience for running diffusion models and other AI models locally using an optimized stack, enabling fast, private, and fully on-device image generation.

<h3 style="color:#44D62C; text-align:left;">🔍 1. Get Model Information</h3>

Use this command to retrieve metadata and configuration details about a specific model from Hugging Face. This helps you understand its architecture, compatibility, license, and intended use before running it.

You can browse compatible models here:  
👉 [vLLM-omni supported model architectures](https://docs.vllm.ai/projects/vllm-omni/en/latest/models/supported_models/#list-of-supported-models-for-nvidia-gpu-amd-gpu)


```bash
rzr-aikit model info Tongyi-MAI/Z-Image-Turbo
```

<h3 style="color:#44D62C; text-align:left;">📥 2. Download a Model</h3>

Use this command to pull a diffusion model from Hugging Face into your local environment.  
This ensures it is available for fast and offline image generation.

```bash
rzr-aikit model download Tongyi-MAI/Z-Image-Turbo
```

<h3 style="color:#44D62C; text-align:left;">🚀 3. Run a Model</h3>

Start a model server to perform local image generation.  
Diffusion models require the `--omni` flag to use the vLLM-omni engine. Note that the optimization stage is skipped for diffusion models.

```bash
rzr-aikit model run Tongyi-MAI/Z-Image-Turbo --omni
```

<h3 style="color:#44D62C; text-align:left;">💬 4. Generate an Image</h3>

Once the model is running, you can send prompts and receive generated images using the `generate` command.  
The generated image will be saved as `output.png` and displayed directly in the terminal.

```bash
rzr-aikit model generate "Sea otter holding a basketball"
cat output.png | display
```

<h3 style="color:#44D62C; text-align:left;">🖥️ 5. (Alternative) Use the Web UI for Image Generation</h3>

As an alternative to the command-line interface, you can use the built-in Web UI to generate images interactively in your browser.  
Once the server is running, open **[http://localhost:7860/](http://localhost:7860/)** to access the interface.

The Web UI exposes a full set of generation parameters for precise control over the output:

| Setting | Description | Default / Example |
|---|---|---|
| **Current Model** | The model currently loaded for inference | `Tongyi-MAI/Z-Image-Turbo` |
| **Prompt** | Describe what you want to generate | `Fluffy cat sitting on glowing Razer keyboard wearing Kraken kitty headphones` |
| **Negative Prompt** | Describe what you want to exclude from the output | `blurry, low quality, distorted` |
| **Height** | Output image height in pixels | `512` |
| **Width** | Output image width in pixels | `512` |
| **Inference Steps** | Number of denoising steps — higher values produce more refined results | `9` |
| **CFG Scale** | Classifier-free guidance scale — controls how closely the output follows the prompt | `0` |
| **Random Seed** | Fixed seed for reproducible results; leave blank to randomize | *(blank)* |

```bash
rzr-aikit ui run
```

<h3 style="color:#44D62C; text-align:left;">🛑 6. Stop the Services</h3>

Stop the Web UI first, then shut down the running model to free GPU and memory resources.

```bash
rzr-aikit ui stop
```

```bash
rzr-aikit model stop
```
