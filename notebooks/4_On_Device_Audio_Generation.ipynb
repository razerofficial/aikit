<h2 style="color:#FFFFFF; text-align:center; font-family:'Roboto', sans-serif; font-size:24px; font-weight:600; letter-spacing:0.08em;">GUIDE FOR USING TTS MODELS FOR LOCAL AUDIO GENERATION.</h2>

<h3 style="color:#44D62C; text-align:left;">Objectives of This Guide</h3>

This guide is designed to walk you through the end-to-end process of generating audio locally using text-to-speech (TTS) models via the `rzr-aikit` interface.

<h3 style="color:#44D62C; text-align:left;">Project Overview</h3>

AIKit is Razer's AI developer environment built to simplify and accelerate machine learning workflows on high-performance Razer hardware. It offers a plug-and-play experience for running TTS models and other AI models locally using an optimized stack, enabling fast, private, and fully on-device audio generation.

<h3 style="color:#44D62C; text-align:left;">🔍 1. Get Model Information</h3>

Use this command to retrieve metadata and configuration details about a specific TTS model from Hugging Face. This helps you understand its architecture, compatibility, license, and intended use before running it.

You can browse compatible models here:  
👉 [vLLM-omni supported model architectures](https://docs.vllm.ai/projects/vllm-omni/en/latest/models/supported_models/#list-of-supported-models-for-nvidia-gpu-amd-gpu)

```bash
rzr-aikit model info mistralai/Voxtral-4B-TTS-2603
```

<h3 style="color:#44D62C; text-align:left;">📥 2. Download a Model</h3>

Use this command to pull a TTS model from Hugging Face into your local environment.  
This ensures it is available for fast and offline audio generation.

```bash
rzr-aikit model download mistralai/Voxtral-4B-TTS-2603
```

<h3 style="color:#44D62C; text-align:left;">🚀 3. Run a Model</h3>

Start a model server to perform local audio generation.  
TTS models require the `--omni` flag to use the vLLM-omni engine.

```bash
rzr-aikit model run mistralai/Voxtral-4B-TTS-2603 --omni
```

<h3 style="color:#44D62C; text-align:left;">💬 4. Generate Audio</h3>

Once the model is running, you can send a text prompt to the `/v1/audio/speech` API endpoint using `curl` and receive generated audio.  
The generated audio will be saved as `output.wav`.

```bash
curl -X POST http://localhost:8000/v1/audio/speech \
    -H "Content-Type: application/json" \
    -d '{
        "model": "mistralai/Voxtral-4B-TTS-2603",
        "input": "Hello! Welcome to Razer AIKit. This is a demonstration of on-device text to speech generation.",
        "voice": "casual_male",
        "language": "English",
        "response_format": "wav",
        "task_type": "CustomVoice"
    }' --output output.wav
```

<h3 style="color:#44D62C; text-align:left;">🔊 5. Play the Audio</h3>

Play the generated audio directly in the notebook using the built-in audio player.
<p style="color:red;">Make sure to switch to a Python kernel (Bash[top-right] -> Select Kernel -> Python 3)</p>

```bash
from IPython.display import Audio, display

display(Audio("output.wav"))
```

<h3 style="color:#44D62C; text-align:left;">🎛️ 6. Advanced: Customize Generation Parameters</h3>

You can fine-tune the audio generation by adjusting the request payload. These parameters are specific to the model being used. Below are some common parameters:

| Parameter | Description | Example |
|---|---|---|
| **input** | The text to synthesize into speech | `"Hello, how are you?"` |
| **voice** | Predefined speaker voice | `"casual_male"`, `"casual_female"` |
| **response_format** | Output audio format | `"wav"` |
| **language** | Target language for synthesis | `"English"`, `"Chinese"`, `"Japanese"` |
| **instructions** | Emotion or style instruction | `"Speak with excitement"` |
| **task_type** | Generation mode | `"CustomVoice"`, `"VoiceDesign"`, `"Base"` |
| **ref_audio** | Reference audio for voice cloning (Base task) | Base64 data URL or HTTP URL |
| **ref_text** | Transcript of reference audio (Base task) | `"This is the reference."` |

**Task Types:**
- **CustomVoice** — Use a predefined speaker voice (default)
- **VoiceDesign** — Generate a voice from a text description via `instructions`
- **Base** — Clone a voice from a reference audio sample

<h3 style="color:#44D62C; text-align:left;">🖥️ 7. (Alternative) Use the Web UI for Audio Generation</h3>

As an alternative to sending requests directly, we recommend using built-in Web UI to generate audio interactively in your browser.  
Once the server is running, open **[http://localhost:7860/](http://localhost:7860/)** to access the interface.

The Web UI provides controls for all generation parameters including voice selection, language, emotion instructions, and voice cloning via reference audio upload.

```bash
rzr-aikit ui run
```

<h3 style="color:#44D62C; text-align:left;">🛑 8. Stop the Services</h3>

Stop the Web UI first (if running), then shut down the running model to free GPU and memory resources.

```bash
rzr-aikit ui stop
```

```bash
rzr-aikit model stop
```
